```python
ciclo = True
while ciclo:

  n = input("Introduce un numero de tres cifras:\n")

  try:
    if len(n) != 3:
      print("El numero debe ser de tres cifras")
      continue

    x = int(n[0])
    y = int(n[1])
    z = int(n[2])

  except ValueError:
    print("El numero debe ser un numero de tres cifras no puede contener letras")
    continue

  if x >= y and x >= z:
    may = x
    if y>=z:
      e = y
      min = z
    else:
      min = y
      e = z
  elif y >= z and y >= x:
    may = y
    if z>=x:
      e = z
      min = x
    else:
      e = x
      min = z
  else:
    may = z
    if y>x:
      e = y
      min = x
    else:
      e = x
      min = y
  print("El numero mas grande es: ",str(may)+str(e)+str(min))

  while True:
    op = input("Quieres introducir otro numero de tres cifras?")
    if op == 'si' or op == 'Si':
      break
    elif op == 'no' or op == 'No':
      ciclo = False
      break
    else:
      print("La respuesta debe ser si o no trata otra vez.")
```
